```python
# OpenAI API 
import os
from llama_index.core import VectorStoreIndex, SimpleDirectoryReader

# 환경변수에서 OpenAI API 키 가져오기
api_key = os.environ.get("OPENAI_API_KEY")

if not api_key:
    raise ValueError("환경변수 OPENAI_API_KEY 가 설정되어 있지 않습니다.")

# 데이터 로드 및 인덱스 생성
documents = SimpleDirectoryReader("data").load_data()  # SimpleDirectoryReader 사용하여 데이터 로드, documents 객체 생성
documents = documents[:5]  # 앞에서 몇 개만 테스트
index = VectorStoreIndex.from_documents(documents)  # VectorStoreIgndex 로 로드한 데이터를 기반으로 인덱스 생성


# 쿼리 엔진 설정 및 실행
query_engine = index.as_query_engine() # as_query_engine() 으로 쿼리 엔진 설정
response = query_engine.query("작가의 청소년 시절은 어땠는지 설명해 줘") # 쿼리 실행 후, 결과 출력
print(response)
```
